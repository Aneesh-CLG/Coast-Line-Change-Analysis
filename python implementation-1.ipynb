```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
from scipy.stats import linregress
from scipy.spatial.distance import directed_hausdorff

# Load grayscale images (2016 and 2020 shorelines)
image_2016 = cv2.imread('Sentinel2_Region_2016_Month_1.png', cv2.IMREAD_GRAYSCALE)
image_2020 = cv2.imread('Sentinel2_Region_2020_Month_1.png', cv2.IMREAD_GRAYSCALE)

# Function to apply simple thresholding for shoreline extraction
def extract_shoreline(image):
    _, binary = cv2.threshold(image, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    contours, _ = cv2.findContours(binary, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    shoreline = np.vstack([c.squeeze() for c in contours if len(c) > 1])  # Stack all shoreline points
    return shoreline

# Extract shorelines from both images
shoreline_2016 = extract_shoreline(image_2016)
shoreline_2020 = extract_shoreline(image_2020)

# Compute Hausdorff Distance (Max shoreline shift)
hausdorff_dist = max(directed_hausdorff(shoreline_2016, shoreline_2020)[0],
                     directed_hausdorff(shoreline_2020, shoreline_2016)[0])

# Convert pixel distance to meters (Assume resolution = 30m per pixel, modify as needed)
pixel_to_meter = 30
shoreline_shift_meters = hausdorff_dist * pixel_to_meter

# **End Point Rate (EPR) Calculation**
years_elapsed = 2020 - 2016  # Time difference
EPR = shoreline_shift_meters / years_elapsed  # meters per year

# **Linear Regression Rate (LRR) Calculation**
years = np.array([2016, 2020])  # Time points
positions = np.array([0, shoreline_shift_meters])  # Shoreline displacement
slope, intercept, r_value, p_value, std_err = linregress(years, positions)
LRR = slope  # meters per year

# **Print Results**
print(f"Hausdorff Distance: {hausdorff_dist:.2f} pixels")
print(f"Shoreline Shift: {shoreline_shift_meters:.2f} meters")
print(f"End Point Rate (EPR): {EPR:.2f} meters/year")
print(f"Linear Regression Rate (LRR): {LRR:.2f} meters/year")

# **Plot Shoreline Shift**
plt.figure(figsize=(8, 5))
plt.scatter(years, positions, color='red', label='Shoreline Positions')
plt.plot(years, slope * years + intercept, color='blue', linestyle='dashed', label=f'LRR Trend (Slope: {LRR:.2f})')
plt.xlabel("Year")
plt.ylabel("Shoreline Position (meters)")
plt.title("Shoreline Change Analysis (2016-2020)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
from scipy.stats import linregress
from scipy.spatial.distance import directed_hausdorff

# Define images and corresponding years manually
images_data = {
    2010: 'Landsat_CloudFree_2010_Month_1.jpg',
    2011: 'Landsat_CloudFree_2011_Month_1.jpg',
    2012: 'Landsat_CloudFree_2012_Month_1.jpg',
    2013: 'Landsat_CloudFree_2013_Month_3.jpg',
    2014: 'Landsat_CloudFree_2014_Month_1.jpg',
    2015: 'Landsat_CloudFree_2015_Month_1.jpg',
    2016: 'Landsat_CloudFree_2016_Month_1.jpg',
    2018: 'Landsat_CloudFree_2018_Month_1.jpg',
    2019: 'Landsat_CloudFree_2019_Month_1.jpg',
}

# Load images manually
images = {}
for year, image_path in images_data.items():
    image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
    if image is not None:
        images[year] = image
    else:
        print(f"Warning: Image for {year} not found!")

# Function to apply simple thresholding for shoreline extraction
def extract_shoreline(image):
    _, binary = cv2.threshold(image, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    contours, _ = cv2.findContours(binary, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    shoreline = np.vstack([c.squeeze() for c in contours if len(c) > 1])  # Stack all shoreline points
    return shoreline

# Extract shorelines for manually inputted images
shorelines = {year: extract_shoreline(images[year]) for year in images}

# Compute Hausdorff Distance between consecutive years
hausdorff_distances = []
shoreline_shifts = []
pixel_to_meter = 30  # Resolution assumption

years = sorted(images.keys())  # Sort years in ascending order

for i in range(len(years) - 1):
    year1, year2 = years[i], years[i + 1]
    if year1 in shorelines and year2 in shorelines:
        hausdorff_dist = max(directed_hausdorff(shorelines[year1], shorelines[year2])[0],
                             directed_hausdorff(shorelines[year2], shorelines[year1])[0])
        hausdorff_distances.append(hausdorff_dist)
        shoreline_shifts.append(hausdorff_dist * pixel_to_meter)
        print(f"{year1}-{year2} Shoreline Shift: {shoreline_shifts[-1]:.2f} meters")

# **End Point Rate (EPR) Calculation**
total_years = years[-1] - years[0]
total_shift = sum(shoreline_shifts)  # Cumulative shift
EPR = total_shift / total_years  # meters per year

# **Linear Regression Rate (LRR) Calculation**
positions = np.cumsum([0] + shoreline_shifts)  # Accumulative positions
slope, intercept, r_value, p_value, std_err = linregress(years[:len(positions)], positions)
LRR = slope  # meters per year

# **Print Results**
print(f"Total Shoreline Shift: {total_shift:.2f} meters")
print(f"End Point Rate (EPR): {EPR:.2f} meters/year")
print(f"Linear Regression Rate (LRR): {LRR:.2f} meters/year")

# **Plot Shoreline Shift with Connected Line**
plt.figure(figsize=(8, 5))
plt.scatter(years[:len(positions)], positions, color='red', label='Shoreline Positions')
plt.plot(years[:len(positions)], positions, color='red', linestyle='-', marker='o', label='Observed Trend')  # Connect points
plt.plot(years[:len(positions)], slope * np.array(years[:len(positions)]) + intercept, 
         color='blue', linestyle='dashed', label=f'LRR Trend (Slope: {LRR:.2f})')

plt.xlabel("Year")
plt.ylabel("Shoreline Position (meters)")
plt.title(f"Shoreline Change Analysis ({years[0]}-{years[-1]})")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import cv2
import numpy as np
import geopandas as gpd
from shapely.geometry import LineString

# Load the processed shoreline image
image_path = "Sentinel2_Region_2016_Month_1.png"  # Change this to your actual file
image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

# Ensure the image is loaded
if image is None:
    raise ValueError(f"Error: Could not open {image_path}")

# Apply Canny Edge Detection
edges = cv2.Canny(image, 50, 150)

# Find contours
contours, _ = cv2.findContours(edges, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# Ensure that at least one contour is found
if not contours:
    raise ValueError("Error: No shoreline detected!")

# Select the largest contour (assuming it represents the shoreline)
largest_contour = max(contours, key=cv2.contourArea)

# Convert the contour points to a valid list of (x, y) tuples
shoreline_coords = [(int(pt[0][0]), int(pt[0][1])) for pt in largest_contour]

# Convert to Shapely LineString
shoreline_line = LineString(shoreline_coords)

# Create a GeoDataFrame
gdf = gpd.GeoDataFrame(geometry=[shoreline_line], crs="EPSG:4326")  # Use WGS84 projection

# Save as a shapefile
gdf.to_file("computed_shoreline.shp")

print("Shoreline shapefile saved successfully!")
```

```python
import cv2
import numpy as np
import geopandas as gpd
from shapely.geometry import LineString

# Load the original shoreline image
image_path = "Landsat_CloudFree_2010_Month_1.jpg"  # Change this to your actual file
image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

# Ensure the image is loaded
if image is None:
    raise ValueError(f"Error: Could not open {image_path}")

# Apply Gaussian Blur to reduce noise
blurred = cv2.GaussianBlur(image, (5, 5), 0)

# Apply adaptive thresholding to enhance contrast
thresholded = cv2.adaptiveThreshold(
    blurred, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 11, 2
)

# Apply morphological operations to remove small gaps
kernel = np.ones((3, 3), np.uint8)
processed = cv2.morphologyEx(thresholded, cv2.MORPH_CLOSE, kernel, iterations=2)

# Detect edges using Canny Edge Detection
edges = cv2.Canny(processed, 50, 150)

# Find contours
contours, _ = cv2.findContours(edges, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# Ensure that at least one contour is found
if not contours:
    raise ValueError("Error: No shoreline detected!")

# Select the largest contour (assuming it represents the shoreline)
largest_contour = max(contours, key=cv2.contourArea)

# Convert the contour points to a list of (x, y) tuples
shoreline_coords = [(int(pt[0][0]), int(pt[0][1])) for pt in largest_contour]

# Optional: Approximate the contour to reduce unnecessary points while maintaining accuracy
epsilon = 0.001 * cv2.arcLength(largest_contour, True)  # Adjust for precision
approx_shoreline = cv2.approxPolyDP(largest_contour, epsilon, False)

# Convert the approximated shoreline points to a list of (x, y) tuples
shoreline_coords = [(int(pt[0][0]), int(pt[0][1])) for pt in approx_shoreline]

# Convert to Shapely LineString
shoreline_line = LineString(shoreline_coords)

# Create a GeoDataFrame
gdf = gpd.GeoDataFrame(geometry=[shoreline_line], crs="EPSG:4326")  # Use WGS84 projection

# Save as a shapefile
gdf.to_file("computed_shoreline_2.shp")

print("Shoreline shapefile saved successfully!")

```

```python
import cv2
import numpy as np
import geopandas as gpd
from shapely.geometry import LineString
import rasterio
from rasterio.transform import Affine

# Load the Sentinel-2 image (Ensure it’s correctly georeferenced)
image_path = "Landsat_CloudFree_2019_Month_1.jpg"
image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

if image is None:
    raise ValueError(f"Error: Could not open {image_path}")

# **1. Multi-Scale Edge Detection**
edges_low = cv2.Canny(image, 30, 100)
edges_high = cv2.Canny(image, 100, 200)
edges = cv2.bitwise_or(edges_low, edges_high)  # Combine both edges for better detection

# **2. Morphological Operations to Refine Edges**
kernel = np.ones((3,3), np.uint8)
edges = cv2.dilate(edges, kernel, iterations=1)
edges = cv2.erode(edges, kernel, iterations=1)

# **3. Contour Detection**
contours, _ = cv2.findContours(edges, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

if not contours:
    raise ValueError("Error: No shoreline detected!")

# **4. Select the Largest Contour (Assuming It's the Shoreline)**
largest_contour = max(contours, key=cv2.contourArea)

# **5. Approximate Contour for Smoothness**
epsilon = 0.001 * cv2.arcLength(largest_contour, True)  # Fine-tune epsilon for detail control
approx_contour = cv2.approxPolyDP(largest_contour, epsilon, True)

# **6. Convert Pixel Coordinates to Georeferenced Coordinates**
def pixel_to_geo(px, py, image_path):
    """ Converts pixel (x, y) to latitude/longitude using image metadata. """
    with rasterio.open(image_path) as src:
        transform = src.transform
        lon, lat = transform * (px, py)
        return lon, lat

shoreline_coords = [pixel_to_geo(int(pt[0][0]), int(pt[0][1]), image_path) for pt in approx_contour]

# **7. Convert to Shapely LineString**
shoreline_line = LineString(shoreline_coords)

# **8. Create a GeoDataFrame and Save as Shapefile**
gdf = gpd.GeoDataFrame(geometry=[shoreline_line], crs="EPSG:4326")
gdf.to_file("computed_shoreline_3.shp")

print("Shoreline shapefile saved successfully with improved accuracy!")
```
